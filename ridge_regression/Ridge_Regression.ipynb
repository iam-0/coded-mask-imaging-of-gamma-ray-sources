```python
# Implying the Ridge Regression method
# min. of c ||Ac - d||^2 + \alpha||c||_2^2
```

```python
import numpy as np
A = np.loadtxt("shadows_matrix.txt")
A = A.T
d = np.loadtxt("dph1.txt")
noise = np.random.poisson(256)
d_noisy = d + noise
print(A.shape)
print(d_noisy.shape)
```

```python
import sklearn as skl
from sklearn import linear_model
from sklearn.linear_model import Ridge

alpha_val = 0.0001 #regularization strength

model = Ridge(alpha=alpha_val, positive=True)
model.fit(A, d_noisy)

c_ridge = model.coef_
```

```python
print(c_ridge.shape)
idx = np.argsort(c_ridge)[::-1]
for i in idx[:10]:
    print(i, c_ridge[i])
```
